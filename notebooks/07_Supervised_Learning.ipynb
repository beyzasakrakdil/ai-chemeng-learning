First, revisit SVM using a dataset with a clearer class structure.

````python
# Create a learnable but imbalanced classification dataset
X, y = make_classification(
    n_samples=1200,
    n_features=3,
    n_informative=2,
    n_redundant=0,
    weights=[0.85, 0.15],
    class_sep=1.2,
    flip_y=0.02,
    random_state=RANDOM_STATE
)

# Keep the test set held out
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=RANDOM_STATE
)

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE
)

results = []

for C_value in [0.01, 0.1, 1, 10, 100]:

    model = Pipeline([
        ("scaler", StandardScaler()),
        ("svm", SVC(
            kernel="linear",
            C=C_value,
            class_weight="balanced"
        ))
    ])

    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring={
            "balanced_accuracy": "balanced_accuracy",
            "fault_precision": "precision",
            "fault_recall": "recall",
            "fault_f1": "f1"
        }
    )

    results.append({
        "C": C_value,
        "Balanced accuracy": scores["test_balanced_accuracy"].mean(),
        "Fault precision": scores["test_fault_precision"].mean(),
        "Fault recall": scores["test_fault_recall"].mean(),
        "Fault F1": scores["test_fault_f1"].mean()
    })

c_results = pd.DataFrame(results)
c_results.round(3)
````

## Kernels and Gamma

A linear SVM tries to separate classes with a straight decision boundary. Some datasets cannot be separated by a single line or plane.

An RBF kernel can create nonlinear decision boundaries by measuring similarities between observations.

### Gamma

Gamma controls how far the influence of an individual training observation extends.

- Small gamma → broad influence and a smoother, simpler decision boundary
- Large gamma → narrow influence and a more curved, complex decision boundary
- Very small gamma → risk of underfitting
- Very large gamma → risk of overfitting

Gamma is relevant to kernels such as RBF, polynomial, and sigmoid. It is not used by a linear kernel.

````python
from sklearn.datasets import make_moons

X_moons, y_moons = make_moons(
    n_samples=1000,
    noise=0.18,
    random_state=RANDOM_STATE
)

X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(
    X_moons,
    y_moons,
    test_size=0.20,
    stratify=y_moons,
    random_state=RANDOM_STATE
)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

models = {
    "Linear SVM": SVC(
        kernel="linear",
        C=1
    ),
    "RBF SVM": SVC(
        kernel="rbf",
        C=1,
        gamma="scale"
    )
}

for ax, (model_name, svm_model) in zip(axes, models.items()):

    pipeline = Pipeline([
        ("scaler", StandardScaler()),
        ("svm", svm_model)
    ])

    pipeline.fit(X_train_m, y_train_m)

    DecisionBoundaryDisplay.from_estimator(
        pipeline,
        X_train_m,
        response_method="predict",
        alpha=0.25,
        ax=ax
    )

    ax.scatter(
        X_train_m[:, 0],
        X_train_m[:, 1],
        c=y_train_m,
        edgecolor="black",
        s=25
    )

    test_accuracy = pipeline.score(X_test_m, y_test_m)

    ax.set_title(
        f"{model_name}\nTest accuracy = {test_accuracy:.3f}"
    )

plt.tight_layout()
plt.show()
````

We create the dataset here:

- `X_moons`: the model inputs, or the two features/coordinates of each point.
- `y_moons`: the class label for each point (`0` or `1`).
- `n_samples=1000`: generates 1,000 observations.
- `noise=0.18`: adds some spread so the data are not perfectly clean. Larger values make the classes overlap more.
- `random_state=RANDOM_STATE`: produces the same dataset each time the cell runs.

### What is the kernel trick?

SVMs naturally work with linear decision boundaries. An RBF kernel lets the model represent the data in a higher-dimensional mathematical space, where a linear separator may be possible.

A common analogy is:

- Interleaved shapes cannot be separated by a straight line in two dimensions.
- Imagine lifting the data into a third dimension.
- A flat surface may separate the shapes there.
- Viewed back in two dimensions, that surface appears as a curved boundary.

The RBF kernel performs the needed similarity calculations without explicitly creating new feature columns. This is called the **kernel trick**.

Because the data form two interleaving half-moons, we expected the RBF model to outperform the linear model. That is what we observed.

This does not mean that RBF is always better than a linear SVM:

- When the true relationship is linear, a linear SVM is simpler and easier to interpret.
- With little data, or with an overly flexible RBF model, the model may memorize the training set.
- The behavior of an RBF SVM depends especially on `C` and `gamma` together.

`C` asks:

> How strongly should classification errors be penalized?

`gamma` asks:

> How wide an area should each training observation influence?

- Small gamma → broad influence and a smooth, simple boundary → risk of underfitting
- Medium gamma → may capture the overall shape of the data
- Large gamma → narrow influence and a highly curved boundary → risk of overfitting

Now we will keep `C=1` fixed and vary only `gamma`.

````python
gamma_values = [0.01, 0.1, 1, 10, 100]

fig, axes = plt.subplots(
    1,
    len(gamma_values),
    figsize=(20, 4),
    sharex=True,
    sharey=True
)

for ax, gamma_value in zip(axes, gamma_values):

    rbf_pipeline = Pipeline([
        ("scaler", StandardScaler()),
        (
            "svm",
            SVC(
                kernel="rbf",
                C=1,
                gamma=gamma_value
            )
        )
    ])

    # Train the model on the training data
    rbf_pipeline.fit(X_train_m, y_train_m)

    # Calculate training and test performance
    train_accuracy = rbf_pipeline.score(X_train_m, y_train_m)
    test_accuracy = rbf_pipeline.score(X_test_m, y_test_m)

    # Plot the decision regions learned by the model
    DecisionBoundaryDisplay.from_estimator(
        rbf_pipeline,
        X_train_m,
        response_method="predict",
        alpha=0.25,
        ax=ax
    )

    # Add the training observations to the plot
    ax.scatter(
        X_train_m[:, 0],
        X_train_m[:, 1],
        c=y_train_m,
        edgecolor="black",
        s=15
    )

    ax.set_title(
        f"gamma = {gamma_value}\n"
        f"Train: {train_accuracy:.3f}\n"
        f"Test: {test_accuracy:.3f}"
    )

plt.tight_layout()
plt.show()
````

### Interpreting the Gamma Results

With small gamma values, the model creates a broad, smooth decision boundary and does not learn the half-moon shape well. As gamma increases, the boundary follows the curved structure more closely.

- `gamma=0.01`: the model is too simple and underfits.
- `gamma=0.1`: the curved structure is partly learned.
- `gamma=1`: the model achieves high test accuracy with a balanced boundary.
- `gamma=10` and `gamma=100`: the boundary becomes more complex, but test accuracy improves only slightly.

In this experiment, higher gamma values did not reduce test performance, so we do not see clear overfitting. However, hyperparameters should be selected with cross-validation, not by inspecting the test set.

````python
# Tune C and gamma together
# C and gamma values to evaluate
parameter_grid = {
    "svm__C": [0.01, 0.1, 1, 10, 100],
    "svm__gamma": [0.01, 0.1, 1, 10, 100]
}

# Put scaling and the RBF SVM in a single pipeline
svm_pipeline = Pipeline([            # A pipeline runs a sequence of transformations and modeling steps.
    ("scaler", StandardScaler()),
    ("svm", SVC(kernel="rbf"))
])

# Five-fold cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE
)

grid_search = GridSearchCV(
    estimator=svm_pipeline,
    param_grid=parameter_grid,
    scoring="accuracy",     # Use accuracy to compare models and select the one with the highest score.
    cv=cv,
    n_jobs=-1,
    return_train_score=True
)

# Search only within the training data
grid_search.fit(X_train_m, y_train_m)

print("Best parameters:", grid_search.best_params_)
print("Best CV accuracy:", round(grid_search.best_score_, 3))

# GridSearchCV refits the best model automatically
best_svm = grid_search.best_estimator_

# Final evaluation on the held-out test data
final_test_accuracy = best_svm.score(X_test_m, y_test_m)

print("Final test accuracy:", round(final_test_accuracy, 3))
````

## k-Nearest Neighbors (KNN)

When a new observation arrives, KNN:

1. Finds the `k` nearest labeled training observations.
2. Checks their class labels.
3. Predicts the majority class among those neighbors.

- Very small `k` → may fit noise; risk of overfitting
- Very large `k` → may lose local patterns; risk of underfitting
- A suitable `k` → balances resistance to noise with the ability to capture local structure

````python
k_values = [1, 5, 25, 100]

fig, axes = plt.subplots(
    1,
    len(k_values),
    figsize=(20, 4),
    sharex=True,
    sharey=True
)

for cur_axes, k_value in zip(axes, k_values):

    knn_pipeline = Pipeline([
        ("scaler", StandardScaler()),
        (
            "knn",
            KNeighborsClassifier(
                n_neighbors=k_value
            )
        )
    ])

    # Fit KNN on the training data
    knn_pipeline.fit(X_train_m, y_train_m)

    # Training and test performance
    train_accuracy = knn_pipeline.score(X_train_m, y_train_m)
    test_accuracy = knn_pipeline.score(X_test_m, y_test_m)

    # Decision regions
    DecisionBoundaryDisplay.from_estimator(
        knn_pipeline,
        X_train_m,
        response_method="predict",
        alpha=0.25,
        ax=cur_axes
    )

    # Actual training observations
    cur_axes.scatter(            # Add the training observations to the current plotting area.
        X_train_m[:, 0],
        X_train_m[:, 1],
        c=y_train_m,
        edgecolor="black",
        s=15
    )

    cur_axes.set_title(
        f"k = {k_value}\n"
        f"Train: {train_accuracy:.3f}\n"
        f"Test: {test_accuracy:.3f}"
    )

plt.tight_layout()  # Prevent the four plots from overlapping.
plt.show() # Display the figure prepared so far.
````

- `fig`: the complete figure, or the large canvas.
- `axes`: the individual plotting areas inside that figure.

In this example:

- `axes[0]` is the first plot.
- `axes[1]` is the second plot.
- `axes[2]` is the third plot.
- `axes[3]` is the fourth plot.

`for ax, k_value in zip(axes, k_values)` pairs each plotting area with a value of `k`.

`ax=ax` tells the plotting function to draw the current model's decision regions in the current plotting area.

The flow is:

1. `plt.subplots()` creates four empty plotting areas.
2. The loop trains a model with a different `k` in each area.
3. It draws the decision regions, adds the observations, and sets a title.
4. `plt.show()` displays the four plots together.

We will choose three KNN settings:

- `n_neighbors`: how many neighbors to consider
- `weights`: whether all neighbors vote equally or closer neighbors have more influence
- `p`: how distance is calculated

In brief:

- `weights="uniform"` → every neighbor has an equal vote
- `weights="distance"` → closer neighbors have a stronger vote
- `p=1` → Manhattan distance
- `p=2` → Euclidean distance

````python
# KNN pipeline
knn_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsClassifier())
])

# KNN parameters to evaluate
knn_parameter_grid = {
    "knn__n_neighbors": list(range(1, 52, 2)),
    "knn__weights": ["uniform", "distance"],
    "knn__p": [1, 2]
}

# Five-fold cross-validation
knn_cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE
)

# Hyperparameter search
knn_grid_search = GridSearchCV(
    estimator=knn_pipeline,
    param_grid=knn_parameter_grid,
    scoring="accuracy",
    cv=knn_cv,
    n_jobs=-1
)

# Search using only the training data
knn_grid_search.fit(X_train_m, y_train_m)

# Best results
print("Best parameters:", knn_grid_search.best_params_)
print(
    "Best CV accuracy:",
    round(knn_grid_search.best_score_, 3)
)

# Best model selected by GridSearchCV
best_knn = knn_grid_search.best_estimator_

# Final result on the held-out test data
knn_test_accuracy = best_knn.score(X_test_m, y_test_m)

print(
    "Final test accuracy:",
    round(knn_test_accuracy, 3)
)
````

This cell performs the following steps:

1. Creates 104 parameter combinations.
2. Evaluates each combination across five folds.
3. Fits the scaler using only the training portion of each fold.
4. Uses KNN to predict the validation portion.
5. Calculates accuracy.
6. Averages the five accuracy scores for each combination.
7. Selects the combination with the highest mean score.
8. Fits the final KNN model on all of `X_train_m` using the selected settings.

````python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    ConfusionMatrixDisplay
)

# Predict on the held-out test set using the best KNN model
y_pred_knn = best_knn.predict(X_test_m)

print("KNN FINAL RESULTS")
print("-" * 40)

print("Best parameters:")
print(knn_grid_search.best_params_)

print(
    "\nMean CV accuracy:",
    round(knn_grid_search.best_score_, 3)
)

print(
    "Final test accuracy:",
    round(accuracy_score(y_test_m, y_pred_knn), 3)
)

print(
    "Final balanced accuracy:",
    round(balanced_accuracy_score(y_test_m, y_pred_knn), 3)
)

print("\nClassification report:")
print(
    classification_report(
        y_test_m,
        y_pred_knn,
        target_names=["Class 0", "Class 1"]
    )
)

ConfusionMatrixDisplay.from_predictions(
    y_test_m,
    y_pred_knn,
    display_labels=["Class 0", "Class 1"],
    cmap="Blues"
)

plt.title("Final KNN Confusion Matrix")
plt.show()
````

Real process data are more challenging. They may include:

- Many sensors
- Irrelevant sensors
- Missing values (`NaN`)
- Sensor drift
- Few fault examples
- Fault types with similar patterns
- Changes in normal operating modes

As the number of features grows, distances between observations can become less informative. This is known as the **curse of dimensionality**. KNN can also be slow on large datasets because it calculates distances to the training data when making predictions.

## Adaptive KNN

### How it differs from standard KNN

**Standard KNN**

Store the training data → Keep the model fixed → Evaluate new observations against the same historical data

**Adaptive KNN**

Store initial normal data → Evaluate a new observation → Add it to the reference data if it is reliably classified as normal → Allow the model to adapt to changing normal conditions

### Main risk: model poisoning

If a new observation is incorrectly classified as normal and added to the reference set, fault behavior can gradually become part of the model's normal region:

Fault observation is mistakenly classified as normal → Added to the reference set → Similar future faults appear more normal → The model gradually degrades

For this reason, we should not automatically add every observation predicted as normal. Common safeguards include:

- Add only observations with high confidence of being normal.
- Pause updates when an alarm is active.
- Use a limited window containing recent observations.
- Check physical safety limits separately.
- Require operator approval when appropriate.

````python
# ============================================================
# CHRONOLOGICAL DATA SPLIT
# ============================================================

# Period boundaries
REFERENCE_END_DAY = 10
CALIBRATION_END_DAY = 15


# First 10 days: initial period of reliable normal operation
reference_mask = (
    time_days <= REFERENCE_END_DAY
)

# Days 10–15: period for calibrating the anomaly threshold
calibration_mask = (
    (time_days > REFERENCE_END_DAY)
    & (time_days <= CALIBRATION_END_DAY)
)

# After day 15: process period to monitor online
stream_mask = (
    time_days > CALIBRATION_END_DAY
)


# Split the 27 process variables into periods
X_reference = X_process.loc[reference_mask].copy()
X_calibration = X_process.loc[calibration_mask].copy()
X_stream = X_process.loc[stream_mask].copy()


# Time values for the same periods
time_reference = time_days.loc[reference_mask].copy()
time_calibration = time_days.loc[calibration_mask].copy()
time_stream = time_days.loc[stream_mask].copy()


# Sanity checks
print("Reference shape:", X_reference.shape)
print(
    "Reference time:",
    round(time_reference.min(), 2),
    "-",
    round(time_reference.max(), 2)
)

print("\nCalibration shape:", X_calibration.shape)
print(
    "Calibration time:",
    round(time_calibration.min(), 2),
    "-",
    round(time_calibration.max(), 2)
)

print("\nStream shape:", X_stream.shape)
print(
    "Stream time:",
    round(time_stream.min(), 2),
    "-",
    round(time_stream.max(), 2)
)

print(
    "\nTotal observations:",
    len(X_reference)
    + len(X_calibration)
    + len(X_stream)
)
````

These are reasonable initial design choices for this experiment:

- First 10 days → approximately 889 observations for the normal reference set.
- Next 5 days → approximately 445 separate normal observations for threshold calibration.
- Remaining 75 days → a long period over which to monitor drift.

Process engineers could adjust these periods based on how long the plant is known to operate reliably under normal conditions.

The key assumption here is that the first 15 days contain no faults and represent healthy operation. If this assumption is wrong, the reference set and threshold may also be unreliable.

````python
# ============================================================
# SCALE THE PROCESS VARIABLES
# ============================================================

process_scaler = StandardScaler()

# Fit the scaler only on the first 10 days of reference data
X_reference_scaled = process_scaler.fit_transform(
    X_reference
)

# Apply the same scaling parameters to the calibration period
X_calibration_scaled = process_scaler.transform(
    X_calibration
)

# Apply the same parameters to the online stream period
X_stream_scaled = process_scaler.transform(
    X_stream
)


print("Reference scaled:", X_reference_scaled.shape)
print("Calibration scaled:", X_calibration_scaled.shape)
print("Stream scaled:", X_stream_scaled.shape)

print(
    "\nOverall reference mean:",
    round(X_reference_scaled.mean(), 4)
)

print(
    "Overall reference standard deviation:",
    round(X_reference_scaled.std(), 4)
)
````

````python
# ============================================================
# BUILD THE STATIC KNN REFERENCE MODEL
# ============================================================

N_NEIGHBORS = 10

static_knn = NearestNeighbors(
    n_neighbors=N_NEIGHBORS,
    metric="euclidean"
)

# Store the normal observations from the first 10 days in the model
static_knn.fit(X_reference_scaled)
# Store the 27-dimensional positions of these 889 normal observations in memory.

print("Static KNN model is ready.")
print("Number of neighbors:", N_NEIGHBORS)
print("Number of reference observations:", len(X_reference_scaled))
````

We fit the scaler only on the reference observations; it learns its parameters from those observations. The same learned scaling parameters are then applied to the calibration and streaming data.

For example, suppose the first 10 days produce these values for a sensor:

- Mean = 8700
- Standard deviation = 30

A reading of 8500 on day 60 is transformed as $z = \frac{8500-8700}{30} = -6.67$. This large negative value indicates that the process is far from its initial state for this sensor.

````python
# ============================================================
# CALCULATE CALIBRATION DISTANCES
# ============================================================
# Use the static KNN model to find the neighbors of the calibration observations.
calibration_distances, calibration_indices = (
    static_knn.kneighbors(X_calibration_scaled)
)

print(
    "Distances shape:",
    calibration_distances.shape
)

print(
    "Indices shape:",
    calibration_indices.shape
)

print("\nDistances for the first calibration observation:")

print(
    np.round(
        calibration_distances[0],
        3
    )
)

print("\nNeighbor indices for the first calibration observation:")

print(
    calibration_indices[0]
)
````

There are 445 calibration observations, and we find 10 neighbors for each one.

`calibration_indices[0]` contains the row indices of the neighbors in the reference table for the first calibration observation.

````python
# Use the calibration distances to determine the anomaly threshold.
anomaly_threshold = np.percentile(calibration_distances.flatten(), 95)
print("Anomaly threshold:", round(anomaly_threshold, 4))
````

### Why we do not use `flatten()` for the threshold

`calibration_distances.flatten()` turns every neighbor distance into one long list, for example:

`[1.2, 1.4, 1.5, ..., 3.0, 1.0, 1.3, ..., 3.4]`

This mixes the small distances to the first, second, and third neighbors with the distance to the tenth neighbor. That can lower the threshold and increase unnecessary alarms.

**The same anomaly score must be used for calibration, threshold selection, and streaming.**

Here, the anomaly score is the distance to the 10th neighbor. Therefore, the threshold must be calculated from `calibration_scores`.

````python
# Use each calibration observation's distance to its 10th-nearest
# neighbor as its anomaly score
calibration_scores = calibration_distances[:, -1]

print(
    "Number of anomaly scores:",
    len(calibration_scores)
)

print(
    "First 10 anomaly scores:"
)

print(
    np.round(
        calibration_scores[:10],
        3
    )
)

print(
    "\nMinimum score:",
    round(calibration_scores.min(), 3)
)

print(
    "Median score:",
    round(np.median(calibration_scores), 3)
)

print(
    "Maximum score:",
    round(calibration_scores.max(), 3)
)
````

This cell is an optional human-readable sanity check. Its output:

- Does not train the model.
- Does not directly determine the threshold.
- Is not required by the next code cell.
- Helps us check that the values look reasonable.

````python
# ============================================================
# DETERMINE THE ANOMALY THRESHOLD
# ============================================================

THRESHOLD_PERCENTILE = 99

anomaly_threshold = np.percentile(
    calibration_scores,
    THRESHOLD_PERCENTILE
)

calibration_alarm_flags = (
    calibration_scores > anomaly_threshold
)
# Create a True/False flag for each calibration observation. True means its score exceeds the anomaly threshold.

print(
    "Threshold percentile:",
    THRESHOLD_PERCENTILE
)

print(
    "Anomaly threshold:",
    round(anomaly_threshold, 4)
)

print(
    "Number of calibration alarms:",
    calibration_alarm_flags.sum()
)

print(
    "Calibration alarm rate:",
    round(
        calibration_alarm_flags.mean(),
        4
    )
)
````

Now we will apply the threshold to the online stream from days 15–90. No faults have been injected yet, so the alarms show how the static model responds to normal process drift.

````python
# ============================================================
# STATIC KNN MONITORING
# ============================================================

# Find the nearest observations in the reference set for each stream observation

stream_distances, stream_indices = (
    static_knn.kneighbors(X_stream_scaled)
)

# Use each observation's distance to its 10th neighbor as its anomaly score
static_stream_scores = stream_distances[:, -1]

# Flag observations whose scores exceed the threshold
static_alarm_flags = (
    static_stream_scores > anomaly_threshold
)

# Convert the time values to a NumPy array
stream_time_values = time_stream.to_numpy()


print("Number of stream observations:", len(static_stream_scores))

print(
    "Number of alarms:",
    static_alarm_flags.sum()
)

print(
    "Alarm rate:",
    round(static_alarm_flags.mean(), 4)
)

if static_alarm_flags.any():

    first_alarm_position = np.where(
        static_alarm_flags
    )[0][0]

    first_alarm_day = stream_time_values[
        first_alarm_position
    ]

    print(
        "First alarm day:",
        round(first_alarm_day, 2)
    )
else:
    print("No alarms occurred during the stream period.")
````

````python
# ============================================================
# ADAPTIVE KNN MONITORING
# ============================================================

from collections import deque

# Approximate number of observations in seven days
ADAPTIVE_WINDOW_SIZE = 620

# Combine the first 15 days of trusted normal history
known_normal_history = np.vstack([
    X_reference_scaled,
    X_calibration_scaled
])

# Initialize the adaptive model with the most recent
# approximately seven days of normal data
adaptive_buffer = deque(
    known_normal_history[-ADAPTIVE_WINDOW_SIZE:], # Normal observations from the latest seven-day window
    maxlen=ADAPTIVE_WINDOW_SIZE # Maximum window size
)

# Empty arrays for storing results
adaptive_stream_scores = np.zeros( # Array for storing anomaly scores
    len(X_stream_scaled)
)

adaptive_alarm_flags = np.zeros( # Array for storing alarm flags
    len(X_stream_scaled), # Number of observations
    dtype=bool
)

# Nearest-neighbor model
adaptive_knn = NearestNeighbors(
    n_neighbors=N_NEIGHBORS,
    metric="euclidean"
)


# Process stream observations in chronological order
for position, new_observation in enumerate( # Iterate over observation positions and new observations
    X_stream_scaled
):

    # Current rolling reference memory
    current_reference = np.asarray( # Retrieve the current reference memory
        adaptive_buffer # rolling reference memory
    )

    # Fit the model to the current reference memory
    adaptive_knn.fit(current_reference) 

    # Calculate distances from the new observation to its neighbors
    new_distances, _ = adaptive_knn.kneighbors(
        new_observation.reshape(1, -1)
    )

    # Distance to the 10th-nearest neighbor
    new_score = new_distances[0, -1]

    adaptive_stream_scores[position] = new_score # Store the new score

    # Check against the threshold
    is_alarm = ( # Determine whether an alarm is raised
        new_score > anomaly_threshold
    )

    adaptive_alarm_flags[position] = is_alarm # Store the new alarm flag

    # Add only observations accepted as normal
    # to the rolling reference memory
    if not is_alarm:
        adaptive_buffer.append(
            new_observation
        )


print(
    "Number of Adaptive KNN alarms:",
    adaptive_alarm_flags.sum()
)

print(
    "Adaptive KNN alarm rate:",
    round(
        adaptive_alarm_flags.mean(),
        4
    )
)

print(
    "Final reference buffer size:",
    len(adaptive_buffer)
)
````

````python
# ============================================================
# COMPARE STATIC AND ADAPTIVE KNN
# ============================================================

comparison_table = pd.DataFrame({
    "Model": [
        "Static KNN",
        "Adaptive KNN"
    ],
    "Alarm count": [
        static_alarm_flags.sum(),
        adaptive_alarm_flags.sum()
    ],
    "Alarm rate": [
        static_alarm_flags.mean(),
        adaptive_alarm_flags.mean()
    ],
    "Median score": [
        np.median(static_stream_scores),
        np.median(adaptive_stream_scores)
    ]
})

comparison_table.round(4)
````

## Controlled Fault Injection

We add controlled faults at known time intervals to a copy of the clean stream. This allows us to evaluate how well Adaptive KNN detects them.

````python
# Keep the clean stream unchanged and add faults to a copy
X_stream_faulty = X_stream.reset_index(drop=True).copy()
fault_time = time_stream.reset_index(drop=True).to_numpy()

# Initially label every observation as normal
fault_type = np.full( 
    len(X_stream_faulty),
    "Normal",
    dtype=object
)


# Return a sensor's standard deviation from the reference period
def get_reference_std(column_name):
    column_position = feature_columns.index(column_name)
    return process_scaler.scale_[column_position]


# =========================================================
# FAULT 1 — Temperature sensor bias
# Furnace_Temp reports a consistently high value from days 30–32
# =========================================================

bias_mask = (
    (fault_time >= 30)
    & (fault_time < 32)
)

temperature_shift = (
    5 * get_reference_std("Furnace_Temp")
)

X_stream_faulty.loc[
    bias_mask,
    "Furnace_Temp"
] += temperature_shift

fault_type[bias_mask] = "Temperature bias"


# =========================================================
# FAULT 2 — Stuck sensor
# Steam_Flow_Coil_3 is stuck at the same value from days 50–52
# =========================================================

stuck_mask = (
    (fault_time >= 50)
    & (fault_time < 52)
)

first_stuck_position = np.where(stuck_mask)[0][0]
last_normal_position = first_stuck_position - 1

stuck_value = X_stream_faulty.loc[
    last_normal_position,
    "Steam_Flow_Coil_3"
]

X_stream_faulty.loc[
    stuck_mask,
    "Steam_Flow_Coil_3"
] = stuck_value

fault_type[stuck_mask] = "Stuck sensor"


# =========================================================
# FAULT 3 — Multivariable process upset
# Hydrocarbon flows decrease from days 70–72,
# while steam flows increase
# =========================================================

upset_mask = (
    (fault_time >= 70)
    & (fault_time < 72)
)

for coil_number in range(1, 7):

    hc_column = f"HC_Flow_Coil_{coil_number}"
    steam_column = f"Steam_Flow_Coil_{coil_number}"

    X_stream_faulty.loc[
        upset_mask,
        hc_column
    ] -= 3 * get_reference_std(hc_column)

    X_stream_faulty.loc[
        upset_mask,
        steam_column
    ] += 3 * get_reference_std(steam_column)

fault_type[upset_mask] = "Process upset"


# Ground-truth fault labels used to evaluate the model
y_fault_true = (fault_type != "Normal").astype(int) 


# Check the number of observations in each fault category
fault_summary = pd.Series(
    fault_type
).value_counts()

print(fault_summary)

print(
    "\nTotal fault observations:",
    y_fault_true.sum()
)

print(
    "Fault rate:",
    round(y_fault_true.mean(), 4)
)
````

We introduced three fault types with different detection challenges:

- **Temperature bias:** The sensor continues to vary but reports values that are consistently shifted in one direction.
- **Stuck sensor:** A failed sensor repeats the same value. This may be harder to detect from a single multivariate snapshot.
- **Process upset:** Several variables change together, indicating a broader change in process behavior. We expect KNN to detect this more readily.

The shifts are set to five or three times each sensor's reference standard deviation. This scales the changes to each sensor's own range instead of adding an arbitrary fixed amount, such as `+100`, to every sensor.

`X_stream.reset_index(drop=True)` resets the row index to:

`0, 1, 2, ..., 6665`

`drop=True` does not remove rows or sensor columns. It discards the old index labels while creating the new index.

```python
fault_time = time_stream.reset_index(drop=True).to_numpy()
```

This line performs three steps:

1. `time_stream` contains the day associated with each observation.
2. `.reset_index(drop=True)` resets the row labels to `0–6665`.
3. `.to_numpy()` converts the Pandas Series into a NumPy array.

````python
# Scale the fault-injected data using parameters learned from the reference period
X_stream_faulty_scaled = process_scaler.transform( # Scale the data because the model operates on scaled features.
    X_stream_faulty 
)


# Reinitialize the adaptive memory
# Do not reuse memory from the previous clean-stream run
known_normal_history = np.vstack([ # np.vstack combines arrays row-wise; here we combine the reference and calibration periods.
    X_reference_scaled,
    X_calibration_scaled
])

fault_test_buffer = deque( # deque supports adding and removing items efficiently; here it holds the adaptive window.
    known_normal_history[-ADAPTIVE_WINDOW_SIZE:],
    maxlen=ADAPTIVE_WINDOW_SIZE
)


# Empty arrays for storing results
fault_test_scores = np.zeros(
    len(X_stream_faulty_scaled) # Store the anomaly score for each stream observation in this array.
)

fault_test_predictions = np.zeros(
    len(X_stream_faulty_scaled), # Store the alarm status (0 or 1) for each stream observation in this array.
    dtype=int
)


# Adaptive KNN model
fault_test_knn = NearestNeighbors(
    n_neighbors=N_NEIGHBORS,
    metric="euclidean"
)


# Process stream observations in order
for position, new_observation in enumerate(
    X_stream_faulty_scaled
):

    # Current memory of normal behavior
    current_reference = np.asarray( # Store the current normal behavior in this array.
        fault_test_buffer
    )

    # Fit KNN to the current memory
    fault_test_knn.fit(current_reference)

    # Calculate distances from the new observation to its neighbors
    new_distances, _ = fault_test_knn.kneighbors(
        new_observation.reshape(1, -1)
    )

    # Distance to the 10th neighbor is the anomaly score
    new_score = new_distances[0, -1]

    fault_test_scores[position] = new_score

    # Set alarm = 1 when the threshold is exceeded
    is_alarm = new_score > anomaly_threshold

    fault_test_predictions[position] = int(is_alarm)

    # Add only observations accepted as normal to memory.
    if not is_alarm:
        fault_test_buffer.append(new_observation)


print(
    "Total alarms raised by the model:",
    fault_test_predictions.sum()
)

print(
    "Model alarm rate:",
    round(fault_test_predictions.mean(), 4)
)
````

````python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)


# Combine all results in one table
adaptive_fault_results = pd.DataFrame({
    "Time_Days": fault_time,
    "Fault type": fault_type,
    "True fault": y_fault_true,
    "Predicted alarm": fault_test_predictions,
    "Anomaly score": fault_test_scores
})


# Overall performance metrics
print("ADAPTIVE KNN CLASSIFICATION REPORT\n")

print(
    classification_report(
        y_fault_true,
        fault_test_predictions,
        target_names=["Normal", "Fault"],
        digits=3,
        zero_division=0
    )
)


# Examine each fault type separately
fault_type_results = (
    adaptive_fault_results[
        adaptive_fault_results["Fault type"] != "Normal"
    ]
    .groupby("Fault type")
    .agg(
        Observation_count=("True fault", "size"),
        Detected_count=("Predicted alarm", "sum"),
        Detection_rate=("Predicted alarm", "mean"),
        Median_score=("Anomaly score", "median"),
        Maximum_score=("Anomaly score", "max")
    )
)

fault_type_results["Detection_rate"] *= 100

print("\nRESULTS BY FAULT TYPE\n")
display(fault_type_results.round(3))


# Confusion matrix
confusion = confusion_matrix(
    y_fault_true,
    fault_test_predictions
)

ConfusionMatrixDisplay(
    confusion_matrix=confusion,
    display_labels=["Normal", "Fault"]
).plot(
    cmap="Blues",
    values_format="d"
)

plt.title("Adaptive KNN — Controlled Fault Test")
plt.show()
````

### Why was the stuck sensor fault mostly missed?

KNN evaluates each observation using the 27 sensor values at that moment. A sensor that repeats the same value for two days can still look normal when each observation is considered on its own.

KNN does not directly ask: **“Has this sensor remained unchanged for a long time?”**

Detecting a stuck sensor requires a time-based rule, for example:

> If a sensor's standard deviation over the last $N$ measurements is approximately zero, raise a stuck-sensor alarm.

The multivariable process upset was detected much more reliably. This shows that a distance-based multivariate detector can recognize coordinated changes across several sensors, while a fault confined to one sensor may need a dedicated temporal check.

## Artificial Neural Networks (ANN)

During each training iteration:

Inputs → Generate a prediction → Compare it with the true answer → Calculate the error (`loss`) → Use backpropagation to trace the error → Update the weights

## Artificial Neural Networks (ANN)

### Purpose

An Artificial Neural Network is a supervised learning method that learns complex, nonlinear relationships between inputs and known outputs.

The model learns a relationship of the form:

$$
X \longrightarrow y
$$

For example, in a chemical plant:

- $X$: temperature, pressure, flow, and other sensor measurements
- $y$: a Normal or Fault label

Using labeled training examples, the model can learn to predict the class of a new sensor observation.

### What are we doing in this example?

We use `make_moons`, a two-feature classification dataset:

- `X_train_m`: the two features for the training observations
- `y_train_m`: the true class labels (`0` or `1`)
- `X_test_m`: test observations not seen during training
- `y_test_m`: the true class labels for the test observations

This is not real plant fault data. It is a controlled visual example for observing how an ANN separates nonlinear classes. If labeled Normal/Fault data were available for a real plant, a similar ANN could be used for supervised fault detection.

### What does an artificial neuron do?

A neuron multiplies input values by weights and adds a bias:

$$
z = w_1x_1 + w_2x_2 + \cdots + w_nx_n + b
$$

- $x_i$: an input or sensor value
- $w_i$: a weight learned during training
- $b$: the bias
- $z$: the weighted sum

The weights represent how the inputs contribute to the prediction. A larger weight can indicate a stronger influence from the corresponding input.

The weighted sum is then passed through an activation function:

$$
a = \text{activation}(z)
$$

For example, the ReLU activation function is:

$$
\text{ReLU}(z) = \max(0,z)
$$

ReLU converts negative values to zero and leaves positive values unchanged. Activation functions allow the network to learn nonlinear, complex relationships.

### Network layers

The ANN used in this example has the following structure:

$$
2 \text{ inputs} \rightarrow 16 \text{ neurons} \rightarrow 8 \text{ neurons} \rightarrow 1 \text{ output}
$$

- **Input layer:** receives the two features.
- **Hidden layers:** learn different combinations of the features.
- **Output layer:** predicts whether an observation belongs to Class 0 or Class 1.

The hidden layers do not contain physical rules that we specify directly. During training, they learn which combinations of inputs help separate the classes.

### How does an ANN learn?

Training repeats the following steps:

1. Pass the training inputs through the network.
2. Generate a class prediction.
3. Compare the prediction with the true class label.
4. Calculate the prediction error (`loss`).
5. Use backpropagation to calculate how much each weight contributed to the error.
6. Update the weights using gradient descent.
7. Repeat until the error decreases sufficiently or a stopping condition is reached.

The basic idea behind a weight update is:

$$
w_{\text{new}} = w_{\text{old}} - \text{learning rate} \frac{\partial \text{loss}}{\partial w}
$$

The `learning rate` controls how much the weights change at each training step.

### Main difference from Adaptive KNN

Adaptive KNN uses the distance between a new observation and known normal observations. An ANN learns the relationship between inputs and target labels from labeled examples.

- Adaptive KNN can be useful when labeled fault data are unavailable.
- An ANN can be a strong fault classifier when enough representative Normal/Fault labels are available.

| Adaptive KNN | ANN |
|---|---|
| May only need normal data | Requires Normal and Fault labels |
| Uses distance to neighboring observations | Learns the relationship between sensors and class labels |
| Can identify new or previously unseen anomalies | Can be effective at recognizing fault patterns represented in training data |
| Can be updated over time | Usually trained with labeled data collected in advance |

Neurons perform the same type of mathematical operation, but they learn different weights. As a result, each neuron can respond to a different pattern in the data.

A single neuron creates one straight decision boundary, which cannot cleanly separate these two half-moon shapes.

### Why use multiple neurons?

Different neurons can learn different lines or parts of a boundary:

- Neuron 1 → learns the left boundary of a region
- Neuron 2 → learns an upper boundary
- Neuron 3 → learns a lower boundary
- Neuron 4 → learns another part of a curve

The model combines these pieces to create a more complex, curved decision boundary.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split


RANDOM_STATE = 42


# Create a nonlinear example dataset with two classes
X_ann, y_ann = make_moons(
    n_samples=1000,
    noise=0.20,
    random_state=RANDOM_STATE
)


# Split the data into training and test sets
X_train_ann, X_test_ann, y_train_ann, y_test_ann = (
    train_test_split(
        X_ann,
        y_ann,
        test_size=0.20,
        stratify=y_ann,
        random_state=RANDOM_STATE
    )
)


print("All X data:", X_ann.shape)
print("All y data:", y_ann.shape)

print("Train X:", X_train_ann.shape)
print("Test X:", X_test_ann.shape)

print(
    "Training class counts:",
    np.unique(
        y_train_ann,
        return_counts=True
    )
)

print(
    "Test class counts:",
    np.unique(
        y_test_ann,
        return_counts=True
    )
)
````

````python
plt.figure(figsize=(8, 5))

plt.scatter(
    X_ann[:, 0],
    X_ann[:, 1],
    c=y_ann,
    cmap="coolwarm",
    edgecolor="black",
    s=25,
    alpha=0.75
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("ANN Classification Data — make_moons")

plt.show()
````

````python
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler


ann_pipeline = Pipeline([ # This pipeline applies scaling and the neural network in sequence.
    (
        "scaler",
        StandardScaler() # Scale the data
    ),
    (
        "ann",
        MLPClassifier( # Artificial Neural Network classifier
            hidden_layer_sizes=(16, 8),
            activation="relu", # ReLU activation function; it allows the network to learn the curved relationship between the two half-moons.
            solver="adam", # Adam is an optimizer that updates the weights to reduce the loss.
            learning_rate_init=0.001,
            max_iter=2000, # Allow up to 2,000 training iterations; the model may stop earlier if it converges.
            random_state=RANDOM_STATE
        )
    )
])


ann_pipeline.fit(
    X_train_ann,
    y_train_ann
)


print("ANN training complete.")
````

`X_train_ann` → `StandardScaler` → Hidden layer with 16 neurons → Hidden layer with 8 neurons → Class 0 or Class 1

````python
trained_ann = ann_pipeline.named_steps["ann"]


print(
    "Number of iterations used:",
    trained_ann.n_iter_
)

print(
    "Final training loss:",
    round(trained_ann.loss_, 4)
)
````

````python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)


# Predictions for the training data
y_train_pred_ann = ann_pipeline.predict(
    X_train_ann
)


# Predictions for test data not seen during training
y_test_pred_ann = ann_pipeline.predict(
    X_test_ann
)


# Calculate accuracy
train_accuracy_ann = accuracy_score(
    y_train_ann,
    y_train_pred_ann
)

test_accuracy_ann = accuracy_score(
    y_test_ann,
    y_test_pred_ann
)


print(
    "Train accuracy:",
    round(train_accuracy_ann, 3)
)

print(
    "Test accuracy:",
    round(test_accuracy_ann, 3)
)

print(
    "Train-test gap:",
    round(
        train_accuracy_ann - test_accuracy_ann,
        3
    )
)


print("\nClassification report:\n")

print(
    classification_report(
        y_test_ann,
        y_test_pred_ann,
        target_names=[
            "Class 0",
            "Class 1"
        ],
        digits=3
    )
)


# Confusion matrix
confusion_ann = confusion_matrix(
    y_test_ann,
    y_test_pred_ann
)

ConfusionMatrixDisplay(
    confusion_matrix=confusion_ann,
    display_labels=[
        "Class 0",
        "Class 1"
    ]
).plot(
    cmap="Blues",
    values_format="d"
)

plt.title("ANN Confusion Matrix")
plt.show()
````

````python
from sklearn.inspection import DecisionBoundaryDisplay


fig, ax = plt.subplots(
    figsize=(9, 6)
)


# Decision regions learned by the ANN
DecisionBoundaryDisplay.from_estimator(
    ann_pipeline,
    X_ann,
    response_method="predict",
    cmap="coolwarm",
    alpha=0.25,
    ax=ax
)


# True test observations
ax.scatter(
    X_test_ann[:, 0],
    X_test_ann[:, 1],
    c=y_test_ann,
    cmap="coolwarm",
    edgecolor="black",
    s=35,
    label="Test observations"
)


# Misclassified test observations
wrong_predictions = (
    y_test_pred_ann != y_test_ann
)

ax.scatter(
    X_test_ann[wrong_predictions, 0],
    X_test_ann[wrong_predictions, 1],
    facecolor="none",
    edgecolor="yellow",
    linewidth=3,
    s=180,
    label="Misclassified"
)


ax.set_xlabel("Feature 1")
ax.set_ylabel("Feature 2")

ax.set_title(
    "ANN Decision Boundary"
)

ax.legend()

plt.show()
````

In the plot:

- Background colors show the decision regions learned by the ANN.
- Point colors show the true classes.
- The two points circled in yellow are misclassified.
- The curved boundary illustrates how multiple neurons can combine different parts of a decision boundary.

````python

ann_architectures = {
    "1 neuron": (1,),
    "4 neurons": (4,),
    "16 + 8 neurons": (16, 8),
    "100 + 100 neurons": (100, 100)
}


ann_architecture_results = []


for architecture_name, architecture in ann_architectures.items():

    model = Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "ann",
            MLPClassifier(
                hidden_layer_sizes=architecture,
                activation="relu",
                solver="adam",
                learning_rate_init=0.001,
                max_iter=2000,
                random_state=RANDOM_STATE
            )
        )
    ])

    model.fit(
        X_train_ann,
        y_train_ann
    )

    train_score = model.score(
        X_train_ann,
        y_train_ann
    )

    test_score = model.score(
        X_test_ann,
        y_test_ann
    )

    trained_model = model.named_steps["ann"]

    ann_architecture_results.append({
        "Architecture": architecture_name,
        "Train accuracy": train_score,
        "Test accuracy": test_score,
        "Train-test gap": train_score - test_score,
        "Iterations": trained_model.n_iter_,
        "Final loss": trained_model.loss_
    })


ann_architecture_results = pd.DataFrame(
    ann_architecture_results
)

ann_architecture_results.round(4)
````

### Effect of ANN Architecture

The networks with one and four hidden neurons showed underfitting. Their training
accuracies remained around 85%, indicating that their capacity was insufficient to
learn the nonlinear decision boundary.

The network with hidden layers `(16, 8)` achieved approximately 97.3% training
accuracy and 99% test accuracy.

Increasing the architecture to `(100, 100)` slightly reduced the training loss but
did not improve test accuracy. Therefore, increasing the number of neurons does
not necessarily improve generalization.

For this dataset, the `(16, 8)` architecture is preferred because it produces the
same test accuracy with substantially fewer trainable parameters and lower model
complexity.

## Random Forest

### How does a Random Forest make its trees different?

#### 1. Bootstrap sampling

Each tree is trained on a random sample drawn from the training data. For example, if the original training set has 800 observations:

- Tree 1 → a random sample of 800 observations
- Tree 2 → a different random sample of 800 observations
- Tree 3 → another random sample of 800 observations

Sampling is done with replacement. An observation can appear more than once in a tree's training sample, while some observations may not appear at all. This is called **bagging** (bootstrap aggregating).

#### 2. Random feature selection

At each split, a tree considers a randomly selected subset of features rather than all features. If there are 27 sensors, for example:

- Tree 1 → temperature, hydrocarbon flow, pressure, and others
- Tree 2 → steam flow, outlet temperature, and others
- Tree 3 → a different combination of sensors

This prevents the trees from becoming identical copies of one another.

### Why might a Random Forest outperform a single tree?

A single Decision Tree:

- Is sensitive to noise.
- Can easily memorize the training data.
- Can change substantially when the data change slightly.

A Random Forest:

- Averages the behavior of many trees.
- Is generally more robust to noise.
- Often generalizes better.
- Can learn complex, nonlinear relationships.
- Can estimate feature importance.

However, it is harder to interpret than one tree, requires more memory and computation, and can become large when it contains many trees.

````python
# Compare a single tree with a forest

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score


# Single decision tree
single_tree = DecisionTreeClassifier(
    random_state=RANDOM_STATE
)


# Random Forest with 200 decision trees
random_forest = RandomForestClassifier(
    n_estimators=200,
    max_features="sqrt",
    random_state=RANDOM_STATE,
    n_jobs=-1
)


# Train both models on the same training data
single_tree.fit(
    X_train_ann,
    y_train_ann
)

random_forest.fit(
    X_train_ann,
    y_train_ann
)


# Training and test predictions
tree_train_prediction = single_tree.predict(
    X_train_ann
)

tree_test_prediction = single_tree.predict(
    X_test_ann
)

forest_train_prediction = random_forest.predict(
    X_train_ann
)

forest_test_prediction = random_forest.predict(
    X_test_ann
)


# Results table
ensemble_results = pd.DataFrame({ # Compare training and test accuracy for a single decision tree and a Random Forest.
    "Model": [
        "Single Decision Tree",
        "Random Forest"
    ],
    "Train accuracy": [
        accuracy_score(
            y_train_ann,
            tree_train_prediction
        ),
        accuracy_score(
            y_train_ann,
            forest_train_prediction
        )
    ],
    "Test accuracy": [
        accuracy_score(
            y_test_ann,
            tree_test_prediction
        ),
        accuracy_score(
            y_test_ann,
            forest_test_prediction
        )
    ]
})


ensemble_results["Train-test gap"] = (
    ensemble_results["Train accuracy"]
    -
    ensemble_results["Test accuracy"]
)


ensemble_results.round(4)
````

````python
from sklearn.model_selection import (
    RepeatedStratifiedKFold,
    cross_validate
)


# Repeat five-fold cross-validation five times
ensemble_cv = RepeatedStratifiedKFold(
    n_splits=5,
    n_repeats=5,
    random_state=RANDOM_STATE
)


models_to_compare = {
    "Single Decision Tree": DecisionTreeClassifier(
        random_state=RANDOM_STATE
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        max_features="sqrt",
        random_state=RANDOM_STATE,
        n_jobs=1
    )
}


cv_results = []


for model_name, model in models_to_compare.items():

    scores = cross_validate(
        model,
        X_ann,
        y_ann,
        scoring="accuracy",
        cv=ensemble_cv,
        n_jobs=-1,
        return_train_score=True
    )

    cv_results.append({
        "Model": model_name,
        "Mean train accuracy": (
            scores["train_score"].mean()
        ),
        "Mean validation accuracy": (
            scores["test_score"].mean()
        ),
        "Validation std": (
            scores["test_score"].std()
        ),
        "Minimum validation accuracy": (
            scores["test_score"].min()
        ),
        "Maximum validation accuracy": (
            scores["test_score"].max()
        )
    })


cv_results = pd.DataFrame(cv_results)

cv_results.round(4)
````

## Model Parameters and Hyperparameters

### Parameters vs. hyperparameters

**Model parameters** are learned by the model during training:

- ANN weights and biases
- Decision Tree split thresholds
- SVM coefficients that define the decision boundary

**Hyperparameters** are settings we choose before training:

- Decision Tree: `max_depth`
- SVM: `C`, `gamma`
- KNN: `n_neighbors`
- ANN: number of neurons, `learning_rate`, `alpha`
- Random Forest: `n_estimators`, `max_depth`

| Model | Regularization approach |
|---|---|
| Decision Tree | Prune the tree or limit its depth |
| SVM | Control `C` and `gamma` |
| KNN | Avoid using an excessively small `k` |
| ANN | Use `alpha`, early stopping, or dropout |
| Random Forest | Limit tree depth or leaf size; average across trees |

Regularization limits unnecessary model complexity and helps prevent the model from memorizing the training data.

**Feature selection** chooses useful features from the columns that already exist.

**Feature engineering** creates new, more informative features from raw columns.
